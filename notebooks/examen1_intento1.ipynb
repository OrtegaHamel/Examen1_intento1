# 2.1 Limpieza de Datos

No se detectaron valores duplicados ni Nulos en el dataset original.  
Sin embargo, se descubre que los valores de la columna 'Total Amount' son de tipo string cuando lo más conveniente es que estén en float.

```python
# Hacemos una copia del Data Frame original para trabajar con ella
df_telco = df.copy()

# Convertimos los datos de la columna 'TotalCharges' a numéricos ya que se encuentran como str
df_telco['TotalCharges'] = pd.to_numeric(df_telco['TotalCharges'], errors='coerce')

# Mostramos los tipos de datos de cada columna del Data Frame para verificar que la conversión se realizó correctamente
print("=" * 50)
print("Tipos de datos:")
print(df_telco.dtypes)

# Mostramos los valores nulos en cada columna del Data Frame después de la conversión
print("=" * 50)
print("Valores nulos en la columna 'TotalCharges' después de la conversión: ")
display(df_telco["TotalCharges"].isnull().sum())
```

Al hacer la conversión aparecen 11 nulos, por lo que se investiga el mejor modo de imputarlos.

```python
# Mostramos los clientes que tienen valores nulos en la columna 'TotalCharges'
print("=" * 50)
print("Cantidad de clientes que tienen valores nulos en la columna 'TotalCharges':")
display(df_telco[df_telco["TotalCharges"].isnull()])
```

Al analizar los resultados, encontramos que los clientes que tienen NaN en 'TotalCharges' coinciden con aquellos que tienen tenure=0, es decir, son los clientes nuevos.  
Ante esto, una ocpión válida sería pensar que el gasto total de estos clientes correspondería al pago del primer mes, es decir, el de la columna 'MonthlyCharges'.

```python
# Rellenamos los valores nulos de la columna 'TotalCharges' con el valor de la columna 'MonthlyCharges'
df_telco.loc[df_telco['TotalCharges'].isnull(), 'TotalCharges'] = df_telco['MonthlyCharges']

# Mostramos los valores nulos en cada columna del Data Frame después de rellenar los valores nulos
print("=" * 50)
print("Valores nulos en la columna 'TotalCharges' después de rellenar los valores nulos:")
display(df_telco["TotalCharges"].isnull().sum())

# Mostramos la cantidad de clientes que tienen valores nulos en la columna 'TotalCharges'
print("=" * 50)
print("Cantidad de clientes que tienen valores nulos en la columna 'TotalCharges':")
display(df_telco[df_telco["TotalCharges"].isnull()])

# Mostramos los clientes que tienen ternure igual a 0 meses para ver cómo queda la columna 'TotalCharges' después de rellenar los valores nulos
print("=" * 50)
print("Clientes que tienen ternure igual a 0 meses:")
display(df_telco[df_telco["tenure"] == 0])
```

# 3.1 Exploración de datos

Para el siguiente gráfico se utilizó como referencia el repositorio de GitHub del Bootcamp

```python
# Histograma de la columna 'tenure' para ver la cantidad de clientes según su tiempo como cliente
fig, ax = plt.subplots(figsize=(10, 5))
ax.hist(df_telco["tenure"], bins=35, color="#6C3CE1", edgecolor="white", alpha=0.85, label="Cantidad de Clientes")

media   = df_telco["tenure"].mean()
mediana = df_telco["tenure"].median()
ax.axvline(media,   color= "#E91E8C", linewidth=2.5, linestyle="--", label=f"Media: {media:.1f}")
ax.axvline(mediana, color="#F59E0B", linewidth=2.5, linestyle=":",  label=f"Mediana: {mediana:.1f}")
ax.set_title('Distribución del tiempo de permanencia de los clientes')
ax.set_xlabel('Tiempo de permanencia (meses)')
ax.set_ylabel('Cantidad de clientes')
ax.legend(fontsize=11)
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()

print("Podemos observar que hay dos picos claros: Muchos clientes nuevos (0-5 meses) y otro grupo de clientes leales (60-30 meses) con pocos clientes en el rango medio")
print("Esto sugiere una alta tasa de abandono entre los clientes nuevos, mientras que los clientes que han estado con la empresa por más tiempo tienden a ser más leales.")
```

Para hacer el siguiente gráfico utilicé ayuda de Google Gemini para lograr el diseño deseado del gráfico de torta

```python
# Gráfico de torta para la columna 'PaymentMethod' para ver la distribución de los métodos de pago entre los clientes
data_pie = df_telco['PaymentMethod'].value_counts()
cmap = plt.get_cmap('Set3')
colores = [cmap(i) for i in range(len(data_pie))]

fig, ax = plt.subplots(figsize=(10, 7))

ax.pie(
    data_pie, 
    labels=data_pie.index, 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=colores,
    explode=[0.03] * len(data_pie)
)

ax.set_title('Distribución de Métodos de Pago', fontsize=15, pad=20)
plt.tight_layout()
plt.show()

print(f"Se observa que el método de pago más común es '{data_pie.idxmax()}' con un {data_pie.max()/len(df_telco)*100:.1f}% de los clientes.")
print("Es importante analizar si los métodos electrónicos están asociados a una mayor tasa de abandono (Churn),")
print("ya que suelen ser menos estables que los pagos automáticos por tarjeta o banco.")
```

Para la realización del siguiente gráfico de boxplots, se utilizó el código disponible en el repositorio de GitHub del Bootcamp

```python
# Box Plot para comparar la distribución de 'tenure' según el 'PaymentMethod'
metodos_pago = df_telco["PaymentMethod"].unique()
datos_box = [df_telco[df_telco["PaymentMethod"] == m]["tenure"].values for m in metodos_pago]
colores = ["#6C3CE1", "#E91E8C", "#4285F4", "#F59E0B"] 

fig, ax = plt.subplots(figsize=(11, 6))
bp = ax.boxplot(datos_box, patch_artist=True, labels=metodos_pago,
                medianprops=dict(color="white", linewidth=2.5))

for patch, color in zip(bp["boxes"], colores):
    patch.set_facecolor(color)
    patch.set_alpha(0.75)

# 5. Personalización estética final
ax.set_title("Distribución de Antigüedad por Método de Pago", fontsize=14, fontweight="bold")
ax.set_ylabel("Antigüedad (meses)", fontsize=12)
ax.set_xlabel("Método de Pago", fontsize=12)
ax.grid(axis="y", alpha=0.25)

plt.tight_layout()
plt.show()

print("Podemos observar que los clientes que utilizan métodos de pago automáticos (Bank transfer y Credit card) tienden a tener una mayor antigüedad (tenure)")
print("en comparación con aquellos que utilizan métodos de pago electrónicos (Mailed check y Electronic check) que son los clientes más nuevos y volátiles.")
```

El gráfico a continuación se basó en un análisis encontrado en la documentación del dataset para ser modificado y aplicado según parámetros propios
https://www.kaggle.com/code/kabure/insightful-eda-churn-customers-models-pipeline

```python
# Relación entre la cantidad de servicios contratados y la tasa de abandono
df_telco['has_internet'] = np.where(df_telco.InternetService != 'No', 'Yes', 'No')
servicios = ['PhoneService', 'OnlineSecurity', 'OnlineBackup', 
             'DeviceProtection', 'TechSupport', 'StreamingTV', 
             'StreamingMovies', 'has_internet']

# Contamos cuántos servicios tiene cada cliente
df_telco['num_services'] = (df_telco[servicios] == 'Yes').sum(axis=1)
counts = df_telco.groupby(['num_services', 'Churn']).size().unstack(fill_value=0)
# Calculamos el % de fuga por cada nivel de servicio
churn_rate = (counts['Yes'] / (counts['Yes'] + counts['No'])) * 100

fig, ax1 = plt.subplots(figsize=(12, 6))
ax2 = ax1.twinx()  # Creamos un segundo eje para el porcentaje

x = counts.index
width = 0.4
ax1.bar(x - width/2, counts['No'], width=width, color="#6C3CE1", alpha=0.7, label="Se quedan")
ax1.bar(x + width/2, counts['Yes'], width=width, color="#F59E0B", alpha=0.8, label="Abandonan")
ax2.plot(x, churn_rate, color="#00C9A7", marker="o", linewidth=3, markersize=8, label="% Tasa de Fuga") # Línea de Tasa de Churn
ax1.set_title("Relación entre Cantidad de Servicios y Lealtad", fontsize=14, fontweight="bold")
ax1.set_xlabel("Número de Servicios Contratados", fontsize=12)
ax1.set_ylabel("Cantidad de Clientes", fontsize=12)
ax2.set_ylabel("Tasa de Fuga (%)", fontsize=12, color="#00C9A7")

# Añadimos etiquetas de porcentaje
for i, rate in enumerate(churn_rate):
    ax2.annotate(f'{rate:.1f}%', (x[i], churn_rate.iloc[i]), 
                 textcoords="offset points", xytext=(0,10), ha='center', 
                 fontsize=10, fontweight='bold', color="#007A65")

ax1.legend(loc="upper right")
ax1.grid(axis="y", alpha=0.2)

plt.tight_layout()
plt.show()

print("Podemos observar que los clientes con 1 sólo servicio tienen una fuga baja.")
print("Con 2 servicios la fuga se dispara al 50%, cifra relevante que indica un punto crítico en la relación entre servicios contratados y abandono.")
print("A partir de 3 servicios, la tasa de fuga baja drásticamente.")
print("Se observa el punto de menor abandono en clientes con 7 y 8 servicios contratados, aunque la cantidad es muy baja en este grupo.")  
print("Esto sugiere que los clientes con pocos servicios son más propensos a abandonar, mientras que aquellos con 3 o más servicios tienden a ser más leales.")
print("Si logramos que un cliente contrate su 3er servicio, su probabilidad de irse cae a la mitad.")
```
