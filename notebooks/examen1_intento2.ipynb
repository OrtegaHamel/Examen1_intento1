### Ajuste de datos
Comenzamos ajustando los tipos de datos de 'release_date' y 'last_update' a datetime

```python
# Hacemos una copia del Data Frame para trabajar con ella
df_vg = df.copy()

# Convertimos los datos de la columna 'release_date' y 'last_update' a datetime
df_vg['release_date'] = pd.to_datetime(df_vg['release_date'], errors='coerce')
df_vg['last_update'] = pd.to_datetime(df_vg['last_update'], errors='coerce')
```

Eliminamos la columna img ya que no estamos trabajando con imágenes. Además, despejamos la visualización de las tablas.

```python
df_vg.drop(columns=['img'], inplace=True)

df_vg.head()
```

### Manejo de valores faltantes
Partiremos con la columna 'developer' que tiene pocos nulos  

```python
# Mostramos los registros donde el desarrollador es nulo
df_vg.loc[(df_vg['developer'].isnull())]
```

```python
# LLenamos los valores nulos de la columna 'developer' con el valor 'Unknown'
df_vg.loc[(df_vg['developer'].isnull()), 'developer'] = 'Unknown'

# Mostramos los registros donde el desarrollador es nulo
df_vg.loc[(df_vg['developer'].isnull())]

# Mostramos los nulos del Data Frame
print("=" * 50)
print("Registros nulos: ")
df_vg.isnull().sum().sort_values(ascending=False)
```

Para la columna release_date que tiene una cantidad considerable de nulos, hay que evitar borrar información relevante. Para ello, buscamos relaciones con otras columnas, como las de ventas.

```python
# Buscamos los NaN de 'release_date' y los NaN de todas las categorías de ventas
df_vg.loc[(df_vg['release_date'].isnull()) & (df_vg['na_sales'].isnull()) & (df_vg['pal_sales'].isnull()) 
& (df_vg['jp_sales'].isnull()) & (df_vg['other_sales'].isnull()) & (df_vg['total_sales'].isnull())]
```

Podemos ver que los NaN funcionan en bloque, por lo que se pueden eliminar las que coincidan con todas las categorías nulas.

```python
# Eliminamos los registros donde 'release_date' es nulo y todas las categorías de ventas son nulas
df_vg = df_vg.loc[~((df_vg['release_date'].isnull()) & (df_vg['na_sales'].isnull()) & (df_vg['pal_sales'].isnull()) 
& (df_vg['jp_sales'].isnull()) & (df_vg['other_sales'].isnull()) & (df_vg['total_sales'].isnull()))]

# Mostramos los nulos del Data Frame después de eliminar los registros
print("=" * 50)
print("Registros nulos: ")
df_vg.isnull().sum().sort_values(ascending=False)
```

Para imputar los nulos que faltan en realease_date usaremos la mediana del año de la consola que corresponda al juego.

```python
# Llenamos los valores nulos de 'release_date' con la mediana del año de la consola correspondiente
df_vg['release_date'] = df_vg['release_date'].fillna(df_vg.groupby('console')['release_date'].transform('median'))

# Mostramos los nulos del Data Frame después de llenar los valores nulos de 'release_date' con la mediana de cada consola
print("=" * 50)
print("Registros nulos: ")
df_vg.isnull().sum().sort_values(ascending=False)
```

Ahora que release_date ya no tiene nulos, podemos imputar la columna last_update con los valores de release_date, ya que suponemos que si el juego no se actualizó, su última versión es la que salió a la venta.

```python
# Llenamos los valores nulos de 'last_update' con la fecha de lanzamiento
df_vg['last_update'] = df_vg['last_update'].fillna(df_vg['release_date'])

# Mostramos los nulos del Data Frame después de llenar los valores nulos de 'last_update' con la fecha de lanzamiento
print("=" * 50)
print("Registros nulos: ")
df_vg.isnull().sum().sort_values(ascending=False)
```

Para los nulos que quedan en las categorías de ventas, decidimos eliminarlos en bloque, ya que no aportan información útil para el análisis de tendencias o éxito comercial.

```python
# Buscamos los NaN de toda la categoría de ventas
df_vg.loc[(df_vg['na_sales'].isnull()) & (df_vg['pal_sales'].isnull()) 
& (df_vg['jp_sales'].isnull()) & (df_vg['other_sales'].isnull()) & (df_vg['total_sales'].isnull())]
```

```python
# Eliminamos los registros donde todas las categorías de ventas son nulas
df_vg = df_vg.loc[~((df_vg['na_sales'].isnull()) & (df_vg['pal_sales'].isnull()) 
& (df_vg['jp_sales'].isnull()) & (df_vg['other_sales'].isnull()) & (df_vg['total_sales'].isnull()))]

# Mostramos los nulos del Data Frame después de eliminar los registros nulos de toda la categoría de ventas
print("=" * 50)
print("Registros nulos: ")
df_vg.isnull().sum().sort_values(ascending=False)
```

Para los valores faltantes de ventas, rellenaremos con el número 0 si el 'total_sale' es 0

```python
cols_sales = ['na_sales', 'pal_sales', 'jp_sales', 'other_sales']
df_vg.loc[df_vg['total_sales'] == 0, cols_sales] = 0

# Mostramos el Data Frame
print("=" * 50)
print("Data Frame: ")
display(df_vg.tail())

# Mostramos los nulos del Data Frame
print("=" * 50)
print("Registros nulos: ")
df_vg.isnull().sum().sort_values(ascending=False)
```

Después de buscar distintas soluciones, optamos por rellenar con 0 los NaN faltantes de las categorías de ventas.

```python
df_vg[cols_sales] = df_vg[cols_sales].fillna(0)

# Mostramos los nulos del Data Frame después de llenar los valores nulos de las categorías de ventas con 0
print("=" * 50)
print("Registros nulos: ")
df_vg.isnull().sum().sort_values(ascending=False)
```

Para la columna critic_score primero buscaremos por título del juego para saber si ya tiene una calificación en otra plataforma y rellenarla con ese valor. Si tiene más de una calificación en distintas plataformas, rellenamos con el promedio.

```python
score_map = df_vg.groupby('title')['critic_score'].mean()
df_vg['critic_score'] = df_vg['critic_score'].fillna(df_vg['title'].map(score_map))

# Mostramos los nulos del Data Frame después de llenar los valores nulos de 'critic_score' con la media de cada título
print("=" * 50)
print("Registros nulos: ")
df_vg.isnull().sum().sort_values(ascending=False)
```

Imputaremos los nulos que quedan con la mediana de critic_score según su género

```python
# Rellenamos con la mediana del género
median_score_by_genre = df_vg.groupby('genre')['critic_score'].transform(lambda x: x.fillna(x.median()))
df_vg['critic_score'] = df_vg['critic_score'].fillna(df_vg['critic_score'].median())

# Mostramos los nulos del Data Frame después de llenar los valores nulos de 'critic_score' con la mediana del género
print("=" * 50)
print("Registros nulos: ")
df_vg.isnull().sum().sort_values(ascending=False)
```

### Multivariables
Para depurar el análisis, observaremos la evolución de las ventas del género sports en las distintas consolas

```python
# Filtramos solo el género Sports y años relevantes
df_sports = df_vg[df_vg['genre'] == 'Sports'].copy()
df_sports['year'] = df_sports['release_date'].dt.year

#Filtramos por las 5 consolas con más ventas en Sports
consoles = df_sports.groupby('console')['total_sales'].sum().nlargest(10).index
df_sports_top = df_sports[df_sports['console'].isin(consoles)]

# Gráfico de líneas multivariado
plt.figure(figsize=(14, 7))
sns.lineplot(data=df_sports_top, x='year', y='total_sales', hue='console', estimator='sum', errorbar=None, marker='o')

plt.title('Evolución de Ventas de Sports por Consola', fontsize=15, fontweight='bold')
plt.xlabel('Año de Lanzamiento')
plt.ylabel('Ventas Totales Anuales (Millones)')
plt.legend(title='Consola', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("=" * 50)
print("La evolución temporal de Sports revela una alta fragmentación de plataformas.")
print("El análisis multivariado permite identificar que el éxito del género no depende de una sola consola, sino que ha sabido adaptarse a múltiples generaciones de hardware.")
print("Se observa cómo el género tiene picos de ventas que coinciden con los ciclos de vida de consolas dominantes, como el auge de la Wii o la PS4.")
```

Llevaremos a cabo un modelo de Regresión Lineal Simple utilizando las ventas del género Sports en Europa como variable predictoria, ya que muestran una fuerte correlación con el mercado global.

```python
from sklearn.linear_model import LinearRegression

# Filtramos el DataFrame para Sports
df_ml_sports = df_vg[df_vg['genre'] == 'Sports'].dropna(subset=['pal_sales', 'total_sales'])

# Definimos las variables
X_sports = df_ml_sports[['pal_sales']]
y_sports = df_ml_sports['total_sales']

# Entrenamos el modelo
modelo_sports = LinearRegression()
modelo_sports.fit(X_sports, y_sports)

# Probamos 5M en Europa (PAL) y predecimos el total
prediccion_sports = modelo_sports.predict(np.array([[5.0]]))


print(f"--- RESULTADOS DEL MODELO (SOLO SPORTS) ---")
print(f"Precisión del modelo (R2): {modelo_sports.score(X_sports, y_sports):.4f}")
print(f"Si un juego de SPORTS vende 5M en Europa, se estima un total de {prediccion_sports[0]:.2f}M.")
```

```python
# Creamos un gráfico de dispersión con la línea de tendencia

# Configuramos el gráfico
plt.figure(figsize=(10, 6))

# Usamos directamente las columnas de df_ml_sports
sns.regplot(x='pal_sales', y='total_sales', data=df_ml_sports, 
            scatter_kws={'alpha':0.3, 'color':'grey'}, 
            line_kws={'color':'#3498db', 'label':'Línea de Predicción (Europa)'})

# Dibujamos la predicción para 5M en Europa
plt.scatter(5, 7.78, color='gold', s=250, marker='*', 
            label='Predicción: 5M Europa -> 7.78M Total', zorder=5, edgecolors='black')

plt.title('Modelo de Regresión: Relación Europa Sales vs Total Sales (SPORTS)', fontsize=14, fontweight='bold')
plt.xlabel('Ventas Europa (PAL) (Millones)')
plt.ylabel('Ventas Totales (Millones)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

Para finalizar este análisis, gracias al alto coeficiente de determinación del modelo, podemos proyectar el éxito de un lanzamiento basándonos en el desempeño inicial en el mercado europeo. Esto convierte a esta región en una herramienta de preventa estratétiga al momento de lanzar un videojuego del género Sport.
